```python
a=130
b=24

if b>a: 
    print(f'{b} is greater than {a}')
else:
    print(f'{b} is less than or equal to {a}')
```

```python
a=410
b=410
if b>a: 
    print(f'{b} is greater than {a}')
elif b==a:
    print(f'{b} is equal to {a}')
else:
    print(f'{b} is less than {a}')
```

## Loops

### While

```python
i=0
while i<4:
    print(i)
    i+=1
```

```python
i=0
while i<10:
    print(2*i+3)
    i+=3
```

### For

```python
for i in range(6):
    print(i)
```

```python
for i in range(10,22,2):
    print(i)
```

```python
cars = ['Tesla','Mazda','Ford','Toyota','Benz']
```

```python
for j in cars:
    print(j)
```
